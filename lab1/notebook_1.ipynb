```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split

# import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(base_path.parent / 'back_dataset' / 'telecom_churn_clean.csv')
df = df.drop(df.columns[0], axis=1)
df.head()
```

```python
df.nunique()
```

```python
# plt.pie(y.value_counts(), labels=['no', 'yes'])
# plt.title("Распределение целевой переменной")
```

```python

X = df.drop(['churn'], axis=1)
y = df['churn']

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.3,
    random_state=123
)
```

```python
import time
from sklearn import svm

start_time = time.perf_counter()

clf1 = svm.SVC(kernel='linear', C=1, random_state=123)
clf1.fit(X_train, y_train)

end_time = time.perf_counter()
training_time = end_time - start_time
print(f"Время обучения: {training_time:.2f} секунд")
print(f"accuracy: {clf1.score(X_test, y_test)}")
```

```python
from sklearn.model_selection import cross_val_score

start_time = time.perf_counter()

clf2 = svm.SVC(kernel='linear', C=1, random_state=123)
scores = cross_val_score(clf2, X, y, cv=5)

end_time = time.perf_counter()
training_time = end_time - start_time
print(f"Время обучения: {training_time:.2f} секунд")
print(f"accuracy: {scores}")
```

```python

from sklearn import preprocessing

scaler = preprocessing.StandardScaler().set_output(transform="pandas")
scaler.fit(X_train)

X_train_scaled = scaler.transform(X_train)
X_test_scaled = scaler.transform(X_test)

X_train_scaled.head()
```

```python

min_max_scaler = preprocessing.MinMaxScaler().set_output(transform="pandas")
X_train_minmax = min_max_scaler.fit_transform(X_train)
X_test_minmax = min_max_scaler.transform(X_test)

X_train_minmax.head()
```

```python

start_time = time.perf_counter()

clf1 = svm.SVC(kernel='linear', C=20, random_state=123)
clf1.fit(X_train_scaled, y_train)

end_time = time.perf_counter()
training_time = end_time - start_time
print(f"Время обучения: {training_time:.2f} секунд")
print(f"accuracy: {clf1.score(X_test_scaled, y_test)}")

y_pred = clf1.predict(X_test_scaled)
from sklearn.metrics import precision_score, recall_score
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
print(f"Precision: {precision:.4f}")
print(f"Recall: {recall:.4f}")

print(np.unique(y_pred))

# start_time = time.perf_counter()

# clf2 = svm.SVC(kernel='linear', C=1, random_state=123)
# scores = cross_val_score(clf2, scaler.transform(X), y, cv=5)

# end_time = time.perf_counter()
# training_time = end_time - start_time
# print(f"Время обучения: {training_time:.2f} секунд")
# print(f"accuracy: {scores}")
# print(f"accuracy: {scores.mean()}")

# y_pred = cross_val_score.predict(X_test_scaled)
# print(np.unique(y_pred))
```
